```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    # model="granite4:350m",
    model='llama3.2:3b',
    validate_model_on_init=True,
    temperature=0,
)
```

```python
from pydantic import BaseModel, Field

class Player(BaseModel):
    """
    Represents a football player with their details.
    """
    name: str = Field(description="The name of the player")
    age: int = Field(description="The age of the player")
    position: str = Field(description="The position the player plays in")

llm_structured = llm.with_structured_output(Player)
```

```python
llm_structured.invoke("Who is the best football player in the world?")
```
