## DictType

```python
from typing import TypedDict

class Movie(TypedDict):
    name: str
    year: int

movie = Movie(name='Lacky number Slevin', year=2005) 
movie
```

```python
from typing import Union

def area(x: Union[int, float], y: Union[int, float]) -> float:
    return x * y 

```

```python
from typing import Optional

def send_message(message: Optional[str]) -> None:
    if message:
        print(f'Send the message {message}')
    else:
        print(f'Send the message : {"Default message"}')

send_message(message=None)
```

```python
from typing import Any

def send_message(message: Any) -> None:
    print(message)
```
